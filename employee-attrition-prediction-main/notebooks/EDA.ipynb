```python
plt.figure(figsize=(6,4))

sns.countplot(data=df, x='Attrition')

plt.title("Employee Attrition")
plt.xlabel("Attrition")
plt.ylabel("Count")

plt.show()
```

Interpretation

A smaller portion left.
This indicates an imbalanced classification problem, so metrics like Precision, Recall, F1-score, and ROC-AUC are more informative than accuracy alone.

```python
#age 
plt.figure(figsize=(8,5))

sns.histplot(df['Age'], bins=20, kde=True)

plt.title("Age Distribution")

plt.show()
```

## Interpretation
Most employees are between 30 and 40 years old.
Very young and older employees are fewer in num

```python
plt.figure(figsize=(8,5))

sns.boxplot(data=df,
            x='Attrition',
            y='Age')

plt.show()
```

## Interpretation
Employees who leave are generally younger.
Younger employees may be more likely to switch jobs for career growth.

```python
#Department vs Attrition
plt.figure(figsize=(8,5))

sns.countplot(data=df,
              x='Department',
              hue='Attrition')

plt.xticks(rotation=20)

plt.show()
```

## Interpretation
sales and research and development departments with the highest number of employees leaving.
These departments may need retention strategies.
